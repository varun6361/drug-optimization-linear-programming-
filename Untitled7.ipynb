```python
import pulp
```

```python
model = pulp.LpProblem("Maximize_Profit", pulp.LpMaximize)
```

```python
x = pulp.LpVariable('Drug_A', lowBound=0)
y = pulp.LpVariable('Drug_B', lowBound=0)
```

```python
model += 20*x + 30*y
```

```python
model += 2*x + y <= 100   # Resource constraint
model += x + 3*y <= 90
```

```python
model.solve()
```

```python
print("Status:", pulp.LpStatus[model.status])

print("Drug A:", x.value())
print("Drug B:", y.value())

print("Maximum Profit:", pulp.value(model.objective))
```
